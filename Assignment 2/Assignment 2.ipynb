```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA
```

```python
df = pd.read_csv("Student_Performance.csv")

df.head()
```

```python
#Task 1

df.info() # Check dataset info

df.isnull().sum() # Check missing values

df.duplicated().sum() # Check duplicate rows

df.describe() # Summary stats
```

```python
#Task 2
df_filled = df.fillna(df.mean(numeric_only=True)) #Missing value strategy 
```

```python
#Task 3

numeric_cols = df.select_dtypes(include=np.number).columns

outliers = {}

for col in numeric_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers[col] = df[(df[col] < lower) | (df[col] > upper)].shape[0]

outliers
for col in numeric_cols:
    
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df[col] = np.clip(df[col], lower, upper)
```

```python
numeric_df = df.select_dtypes(include=np.number)
minmax_scaler = MinMaxScaler()

df_minmax = pd.DataFrame(
    minmax_scaler.fit_transform(numeric_df),
    columns=numeric_df.columns
)

df_minmax.head()
```

```python
#Z-Score normalization
scaler = StandardScaler()

df_zscore = pd.DataFrame(
    scaler.fit_transform(numeric_df),
    columns=numeric_df.columns
)

df_zscore.head()
```

```python
#Task 5 - Apply PCA and Interpret Explained Variance
X = df_zscore

pca = PCA()

pca.fit(X)

explained_variance = pca.explained_variance_ratio_

explained_variance
```
