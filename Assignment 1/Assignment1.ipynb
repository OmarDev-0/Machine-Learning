```python
# summary of stats
df.describe(include='all')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['study_hours'], bins=20)
plt.title("exam_score")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='study_hours', hue='academic_level', bins=15, multiple="stack")
plt.title("Study Hour Distribution by Academic Level")
plt.xlabel("Hours Spent Studying")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='sleep_hours', hue='gender', bins=10, kde=True, element="step")
plt.title("Distribution of Sleep Hours by Gender")
plt.show()
```

```python
country_revenue = df.groupby('Country')['Amount'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
country_revenue.plot(kind='bar')
plt.title("Total Revenue by Country")
plt.ylabel("Revenue")
plt.show()

country_revenue
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='study_hours', y='exam_score', data=df, hue='academic_level', alpha=0.7)
plt.title("Study Hours vs. Exam Score")
plt.xlabel("Weekly Study Hours")
plt.ylabel("Exam Score (%)")
plt.show()
```

```python
plt.figure(figsize=(10, 8))
academic_corr = df[['study_hours', 'self_study_hours', 'online_classes_hours', 'focus_index', 'exam_score']].corr()

sns.heatmap(academic_corr, annot=True, cmap='RdYlGn', center=0)
plt.title("Correlation: Study Habits vs. Performance")
plt.show()
```

```python
# Converts date column to a monthly period
df['Month'] = df['Date'].dt.to_period('M')

# Groups by month and calculate the average burnout
monthly_burnout = df.groupby('Month')['burnout_level'].mean()

plt.figure(figsize=(10,5))
monthly_burnout.plot(marker='o', color='red', linestyle='-')
plt.title("Average Student Burnout Trend")
plt.ylabel("Burnout Level")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
